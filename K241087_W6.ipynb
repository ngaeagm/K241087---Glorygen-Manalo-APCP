```python
import time

start = time.time()

text = "Kent Institute Australia"

# Split into three substrings
parts = text.split()   # ['Kent', 'Institute', 'Australia']

# Count lengths
len1 = len(parts[0])
len2 = len(parts[1])
len3 = len(parts[2])

print("Substring 1:", parts[0], "Length:", len1)
print("Substring 2:", parts[1], "Length:", len2)
print("Substring 3:", parts[2], "Length:", len3)

end = time.time()
print("Execution time:", end - start, "seconds")
```
