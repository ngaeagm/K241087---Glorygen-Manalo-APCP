```python
# Superclass: Student
class Student:
    def __init__(self, student_id, name):
        self.__student_id = student_id
        self.__name = name

    def get_student_id(self):
        return self.__student_id

    def get_name(self):
        return self.__name

    def show_details(self):
        print(f"Student ID: {self.__student_id}")
        print(f"Name: {self.__name}")


# Subclass: IT Student
class ITStudent(Student):
    def __init__(self, student_id, name, programming_language):
        super().__init__(student_id, name)
        self.__programming_language = programming_language

    def show_details(self):
        super().show_details()
        print("Major: Information Technology")
        print(f"Programming Language: {self.__programming_language}")


# Subclass: Business Student
class BusinessStudent(Student):
    def __init__(self, student_id, name, specialization):
        super().__init__(student_id, name)
        self.__specialization = specialization

    def show_details(self):
        super().show_details()
        print("Major: Business")
        print(f"Specialization: {self.__specialization}")


# Subclass: Accounting Student
class AccountingStudent(Student):
    def __init__(self, student_id, name, certification_goal):
        super().__init__(student_id, name)
        self.__certification_goal = certification_goal

    def show_details(self):
        super().show_details()
        print("Major: Accounting")
        print(f"Certification Goal: {self.__certification_goal}")


# --- Testing the classes ---

# details: IT Student
glorygen = ITStudent("K241087", "Glorygen Manalo", "Python")

# Other sample students
business = BusinessStudent("K00000", "Maria Santos", "Marketing")
accounting = AccountingStudent("K00001", "John Cruz", "CPA")

print("\n--- IT Student (Glorygen) ---")
glorygen.show_details()

print("\n--- Business Student ---")
business.show_details()

print("\n--- Accounting Student ---")
accounting.show_details()
```
