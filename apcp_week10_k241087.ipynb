```python
def factorial(n):
    # Base case
    if n == 0:
        return 1
    else:
        # Recursive case
        return n * factorial(n - 1)

# Ask user for input
num = int(input("Enter a number: "))

# Calculate factorial
result = factorial(num)

print(f"The factorial of {num} is {result}")
```
