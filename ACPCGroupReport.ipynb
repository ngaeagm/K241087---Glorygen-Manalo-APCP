```python
import pandas as pd

df = pd.read_csv('/content/vic_traffic_clean.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Violation Type', palette='viridis')
plt.title('Number of Violations by Type')
plt.xticks(rotation=45)
plt.show()
```

```python
payment_counts = df['Payment Status'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%', colors=['#4CAF50','#FF5252'])
plt.title('Payment Status Distribution')
plt.show()
```

```python
monthly_counts = df.groupby('Violation Month').size()

plt.figure(figsize=(10,6))
monthly_counts.plot(kind='line', marker='o', color='blue')
plt.title('Trend of Violations by Month')
plt.xlabel('Month')
plt.ylabel('Number of Violations')
plt.grid(True)
plt.show()
```
