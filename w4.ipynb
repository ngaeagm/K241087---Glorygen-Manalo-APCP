```python
# Grade functions
def grade_A(score):
    return score >= 85

def grade_B(score):
    return 70 <= score < 85

def grade_C(score):
    return 60 <= score < 70

def grade_D(score):
    return 50 <= score < 60

def grade_F(score):
    return score < 50

# Function to determine grade using the above functions
def get_grade(score):
    if grade_A(score):
        return "A"
    elif grade_B(score):
        return "B"
    elif grade_C(score):
        return "C"
    elif grade_D(score):
        return "D"
    else:
        return "F"

# Ask user for score
score = float(input("Enter your score: "))

# Determine grade
final_grade = get_grade(score)

# Write result to a file
with open("grade_result.txt", "w") as file:
    file.write(f"Score: {score}\n")
    file.write(f"Final Grade: {final_grade}\n")

print("Grade has been written to grade_result.txt")
```
