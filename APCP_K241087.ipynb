```python
import time

start_time = time.time()

# Grade Calculator Program
score = float(input("Enter your percentage score: "))

if score >= 85 and score <= 100:
    grade = "High Distinction (HD)"
elif score >= 75 and score <= 84:
    grade = "Distinction (D)"
elif score >= 65 and score <= 74:
    grade = "Credit (CR)"
elif score >= 50 and score <= 64:
    grade = "Pass (P)"
elif score >= 0 and score <= 49:
    grade = "Fail (N or F)"
else:
    grade = "Invalid score. Please enter a value between 0 and 100."

print("Glorygen, your grade is:", grade)

end_time = time.time()
print("Execution time:", end_time - start_time, "seconds")

```
